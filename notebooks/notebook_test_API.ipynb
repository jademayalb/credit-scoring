### Introduction

Ce notebook permet de tester l’API de scoring crédit développée et déployée dans le cadre du projet MLOps.  
L’API, hébergée sur Heroku, retourne la probabilité de défaut de paiement ainsi qu’une décision d’octroi ou non de crédit à partir de l’identifiant d’un client.

**Objectifs de ce notebook :**
- Vérifier que l’API répond correctement pour un client existant (retourne la prédiction attendue)
- Vérifier la gestion d’un client inexistant (retourne une erreur appropriée)
- Illustrer l’intégration de l’API dans une démarche de validation MLOps

**URL de l’API testée :**  
[https://credit-scoring-jademayalb-db8bcc609fed.herokuapp.com/predict/{client_id}](https://credit-scoring-jademayalb-db8bcc609fed.herokuapp.com/predict/{client_id})

```python
# Imports
import requests
import pandas as pd
```

## Test pour un client existant

Nous allons interroger l’API pour l’identifiant client `100091`.

```python
url = 'https://credit-scoring-jademayalb-db8bcc609fed.herokuapp.com/predict/100091'
response = requests.get(url)
print(f"Status code : {response.status_code}")
result = response.json()
print(result)
```

## Test pour un client inexistant

Nous allons interroger l’API pour un identifiant client qui n’existe pas (`999999999`).

```python
url_inexistant = 'https://credit-scoring-jademayalb-db8bcc609fed.herokuapp.com/predict/999999999'
response_inexistant = requests.get(url_inexistant)
print(f"Status code : {response_inexistant.status_code}")
result_inexistant = response_inexistant.json()
print(result_inexistant)
```

## 4. Analyse des réponses

- **Client existant :** on doit obtenir un code 200, avec une réponse contenant la probabilité de défaut et la décision.
- **Client inexistant :** on doit obtenir un code 404, avec un statut `"NOT_FOUND"`.

On peut également afficher la réponse dans un DataFrame pour une meilleure lisibilité.

```python
# Pour le client existant
pd.DataFrame([result])
```

```python
# Pour le client inexistant
pd.DataFrame([result_inexistant])
```

## 5. Conclusion

Ce notebook démontre que l’API répond de façon appropriée selon que le client existe ou non, et permet de valider son intégration dans le pipeline MLOps.
